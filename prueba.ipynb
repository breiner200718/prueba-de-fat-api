# **ejercicio 1**
#1.Registrar la siguiente info mediante una funcion

 *tipo de documento
 *documento
 *nombre
 *apellidos
 *ficha
 *programa

```python
def registrar_estudiante(tipo_doc, documento, nombre, apellidos, ficha, programa):

    estudiante = {
        "tipo_documento": tipo_doc,
        "documento": documento,
        "nombre": nombre,
        "apellidos": apellidos,
        "ficha": ficha,
        "programa": programa
    }

    return estudiante


# Lista donde se guardarán todos los estudiantes
estudiantes = []

while True:
    print("\n=== REGISTRO DE ESTUDIANTE ===")

    tipo_doc = input("Tipo de documento (CC/TI): ")
    documento = input("Número de documento: ")
    nombre = input("Nombre: ")
    apellidos = input("Apellidos: ")
    ficha = input("Número de ficha: ")
    programa = input("Programa: ")

    alumno = registrar_estudiante(
        tipo_doc,
        documento,
        nombre,
        apellidos,
        ficha,
        programa
    )

    estudiantes.append(alumno)

    continuar = input("\n¿Desea registrar otro estudiante? (S/N): ").upper()

    if continuar != "S":
        break

print("\n=== LISTA DE ESTUDIANTES REGISTRADOS ===")
for i, estudiante in enumerate(estudiantes, start=1):
    print(f"\nEstudiante {i}")
    print(f"Tipo de documento: {estudiante['tipo_documento']}")
    print(f"Documento: {estudiante['documento']}")
    print(f"Nombre: {estudiante['nombre']}")
    print(f"Apellidos: {estudiante['apellidos']}")
    print(f"Ficha: {estudiante['ficha']}")
    print(f"Programa: {estudiante['programa']}")


```
